```python
import pandas as pd

df = pd.read_csv("/content/session4_dummy_students_data.csv.xls")
print(df)
```

```python
sci= df['science'].to_numpy()
math = df['math'].to_numpy()
english = df['english'].to_numpy()
df['average'] = (sci + math + english )/3
```

```python
filter_1 = df[df['average']>85]
filter_2 = df[df['attendance_percent'] < 75 ]
filter_3 = df[df['city']=='Mathura']
print(filter_1)
print(filter_2)
print(filter_3)
```

```python
for i in df['average']:
  if i >=40:
    df['result'] = 'Pass'
  else :
    df['result'] = 'Fail'
print(df)
```

```python
print(df['average'])
print(df['average'].max())
print(df['average'].min())
print(df[df['average']>=90]['name'])
print(df[df['attendance_percent']<75]['name'])
print(df[df['assignments_submitted']<6]['name'])
```

```python
print(df['final_project_score'].mean())
print(df[df['study_hours_per_week']>8]['name'])
mumbai = df[df['city']=='Mumbai']['average'].mean()
pune = df[df['city']=='Pune']['average'].mean()
if mumbai > pune :
  print('Mumbai > Pune')
elif mumbai == pune :
  print('Mumbai = Pune')
else  :
  print('Mumbai < Pune')
```
