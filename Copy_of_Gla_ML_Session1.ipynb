Calculate the avg marks and prin t class fialed or passed is the avg is 50 or above

```python
marks = [45, 60, 72, 38, 90]

total = 0

for i in range(len(marks) - 1):
    total += marks[i]

average = total / len(marks)

if average > 50:
    print("Class passed")
else:
    print("Class failed")

print("Average:", average)
```

```python
import pandas as pd
```

```python
data = {
    "student_id": [101, 102, 103, 104, 105, 106],
    "attendance_percent": [92, 67, 81, 45, 74, 88],
    "assignment_score": [18, 12, 15, 8, 14, 19],
    "quiz_score": [72, 48, 65, 30, 55, 80],
    "lab_completed": [True, False, True, False, True, True]
}

df = pd.DataFrame(data)

df
```

```python
df.shape
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df['total_score'] = df['assignment_score'] + df['quiz_score']
df
```

```python
df['eligible'] = (
    (df['attendance_percent'] >= 75) &
    (df['total_score'] >= 70) &
    (df['lab_completed'])
)
df
```

```python
eligible_students = df[df["eligible"] == True]
eligible_students
```

```python
import matplotlib.pyplot as plt

plt.hist(df['total_score'],bins=10)
plt.title("Histogram of Total Scores")
plt.xlabel("Total Score")
plt.ylabel("Frequency")
plt.show()



plt.boxplot(df['total_score'])
plt.title("Boxplot")
plt.ylabel("Total Score")
plt.show()



plt.violinplot(df['total_score'])
plt.title("violin plot")
plt.ylabel("Total Score")
plt.show()


plt.plot(df['total_score'])
plt.title("lineplot")
plt.xlabel('Index')

plt.ylabel('total score')
plt.show()



plt.scatter(df.index,df['total_score'])
plt.title('scatter plot')
plt.xlabel('index')
plt.ylabel('total score')

plt.show()




score_count = df['total_score'].value_counts().sort_index()
plt.bar(score_count.index,score_count.values)
plt.title("bar graph")
plt.xlabel("total score")
plt.ylabel("count")
plt.show()
```
